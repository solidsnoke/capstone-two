# Capstone Project - Modeling Carbon Dioxide Emissions


This notebook focuses on building machine learning models to predict carbon dioxide emissions (`co2`) based on various economic, environmental, and demographic features.

### Objective:
- Train 2–3 regression models
- Tune hyperparameters
- Evaluate and compare model performance
- Select the best-performing model


## 1. Load and Prepare Data


We begin by loading the cleaned dataset, removing unneeded columns (`iso_code`, `co2_change`), and converting the categorical `country` column into numeric form using one-hot encoding.


```

import pandas as pd

# Load data
df = pd.read_csv("cleaned_co2_data.csv")

# Drop identifier and redundant columns
df_model = df.copy()
target = 'co2'
cols_to_drop = ['iso_code', 'co2_change']
df_model = df_model.drop(columns=cols_to_drop)

# Convert country into dummy variables
df_model = pd.get_dummies(df_model, columns=['country'], drop_first=True)

# Split into features and target
X = df_model.drop(columns=[target])
y = df_model[target]
```

## 2. Impute and Scale Data


Missing values in the dataset are imputed using the **mean strategy**, and features are standardized using **StandardScaler**. This step ensures that all features are on the same scale, which is important for models that are sensitive to feature magnitudes.


```

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# Impute missing values
imputer = SimpleImputer(strategy='mean')
X_imputed = imputer.fit_transform(X)

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_imputed)

# Optional: convert back to DataFrame for readability
X_scaled = pd.DataFrame(X_scaled, columns=X.columns)
```

## 3. Train/Test Split


We split the dataset into training and test sets (80/20 split) to evaluate model performance on unseen data.


```

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
```

## 4. Model 1: Linear Regression


Linear Regression serves as a baseline model for regression problems. It is simple, interpretable, and fast to train.


```

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

lr = LinearRegression()
lr.fit(X_train, y_train)
y_pred_lr = lr.predict(X_test)

mae_lr = mean_absolute_error(y_test, y_pred_lr)
rmse_lr = np.sqrt(mean_squared_error(y_test, y_pred_lr))
r2_lr = r2_score(y_test, y_pred_lr)

print("Linear Regression:")
print(f"MAE: {mae_lr:.2f}, RMSE: {rmse_lr:.2f}, R2: {r2_lr:.2f}")
```

## 5. Model 2: Random Forest Regressor


Random Forest is a tree-based ensemble model that often performs well out-of-the-box. We also apply **GridSearchCV** to tune key hyperparameters.


```

from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV

rfr = RandomForestRegressor(random_state=42)
param_grid_rf = {
    'n_estimators': [50, 100],
    'max_depth': [None, 10],
}
grid_rf = GridSearchCV(rfr, param_grid_rf, cv=5, scoring='neg_mean_squared_error', n_jobs=-1)
grid_rf.fit(X_train, y_train)
y_pred_rf = grid_rf.predict(X_test)

mae_rf = mean_absolute_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_rf))
r2_rf = r2_score(y_test, y_pred_rf)

print("Random Forest Regressor:")
print(f"Best Params: {grid_rf.best_params_}")
print(f"MAE: {mae_rf:.2f}, RMSE: {rmse_rf:.2f}, R2: {r2_rf:.2f}")
```

## 6. Model 3: XGBoost Regressor


XGBoost is a powerful boosting model known for its accuracy and efficiency. We use cross-validation to tune it just like with Random Forest.


```

from xgboost import XGBRegressor

xgb = XGBRegressor(random_state=42)
param_grid_xgb = {
    'n_estimators': [50, 100],
    'max_depth': [3, 5],
}
grid_xgb = GridSearchCV(xgb, param_grid_xgb, cv=5, scoring='neg_mean_squared_error', n_jobs=-1)
grid_xgb.fit(X_train, y_train)
y_pred_xgb = grid_xgb.predict(X_test)

mae_xgb = mean_absolute_error(y_test, y_pred_xgb)
rmse_xgb = np.sqrt(mean_squared_error(y_test, y_pred_xgb))
r2_xgb = r2_score(y_test, y_pred_xgb)

print("XGBoost Regressor:")
print(f"Best Params: {grid_xgb.best_params_}")
print(f"MAE: {mae_xgb:.2f}, RMSE: {rmse_xgb:.2f}, R2: {r2_xgb:.2f}")
```

## 7. Model Comparison and Selection


We compare the models on key metrics: **MAE**, **RMSE**, and **R²**. The final model can then be selected based on the balance of performance and interpretability.


```

results = pd.DataFrame({
    'Model': ['Linear Regression', 'Random Forest', 'XGBoost'],
    'MAE': [mae_lr, mae_rf, mae_xgb],
    'RMSE': [rmse_lr, rmse_rf, rmse_xgb],
    'R2 Score': [r2_lr, r2_rf, r2_xgb]
})

print("Model Comparison:")
print(results.sort_values(by='RMSE'))
```
